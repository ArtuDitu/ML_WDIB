# 📚 Część 1: Wprowadzenie teoretyczne

## Co to jest uczenie maszynowe?
Wyobraź sobie, że uczysz małe dziecko odróżniać jabłka od gruszek. 🍎🍐

Nie dajesz mu listy reguł w stylu: *„jeśli owoc jest okrągły, czerwony i ma ogonek krótszy niż 2 cm, to jest to jabłko”*.
Zamiast tego **pokazujesz mu wiele przykładów** i mówisz: *„to jest jabłko”*, *„a to gruszka”*.
Po jakimś czasie dziecko **samo** rozpoznaje owoce, których nigdy wcześniej nie widziało.

Dokładnie tak działa **uczenie maszynowe**: komputer uczy się **na przykładach (danych)**, a nie z reguł spisanych ręcznie przez programistę.

> 💬 Arthur Samuel (1959), pionier sztucznej inteligencji: uczenie maszynowe to dziedzina, która daje komputerom
> umiejętność uczenia się **bez bycia wprost zaprogramowanym**.

## Klasyczne programowanie vs. uczenie maszynowe

| | 🧮 Klasyczne programowanie | 🤖 Uczenie maszynowe |
|---|---|---|
| Co dajemy komputerowi? | **reguły** (kod) + dane | **dane** + poprawne odpowiedzi |
| Co dostajemy? | odpowiedzi | **reguły**, czyli **model** |
| Przykład | kalkulator podatku | filtr antyspamowy w poczcie |

## AI, ML, Deep Learning – co jest czym?
Pomyśl o rosyjskiej matrioszce 🪆:
- **Sztuczna inteligencja** (*Artificial Intelligence, AI*) – największa lalka: wszystko, co sprawia, że komputer zachowuje się „inteligentnie”.
- **Uczenie maszynowe** (*Machine Learning, ML*) – lalka w środku: AI, które **uczy się z danych**.
- **Uczenie głębokie** (*Deep Learning, DL*) – najmniejsza lalka: ML oparte na dużych sieciach neuronowych.

## Trzy rodzaje uczenia maszynowego

### 1. Uczenie nadzorowane (*Supervised Learning*) 👩‍🏫
Jak nauka z nauczycielem, który ma **klucz odpowiedzi**. Każdy przykład w danych ma podaną poprawną odpowiedź (**etykietę**, *label*).
- **Regresja** (*Regression*) – przewidujemy **liczbę**: cenę mieszkania, koszt ubezpieczenia, jutrzejszą temperaturę.
- **Klasyfikacja** (*Classification*) – przewidujemy **kategorię**: spam / nie spam, gatunek kwiatu, chory / zdrowy.

### 2. Uczenie nienadzorowane (*Unsupervised Learning*) 🧩
Nie ma nauczyciela ani odpowiedzi. Model **sam szuka struktury** w danych – jak dziecko, które sortuje klocki LEGO po kolorach, choć nikt mu nie kazał.
- **Grupowanie** (*Clustering*) – np. podział klientów sklepu na grupy o podobnych zwyczajach.
- **Wykrywanie anomalii** (*Anomaly Detection*) – np. wyłapywanie podejrzanych transakcji kartą.

### 3. Uczenie ze wzmocnieniem (*Reinforcement Learning*) 🐕
Jak tresura psa: za dobre zachowanie jest nagroda (smakołyk), za złe – brak nagrody. Agent uczy się metodą prób i błędów.
Tak trenuje się np. programy do gier czy roboty. **PyCaret tego nie obsługuje**, więc na kursie tylko o tym wspominamy.

## Najważniejsze pojęcia na przykładzie tabeli
Dane w ML to zwykle **tabela** (jak w Excelu). Przykład – mieszkania na sprzedaż:

| metraż (m²) | liczba pokoi | dzielnica | **cena (zł)** |
|---|---|---|---|
| 45 | 2 | Mokotów | **620 000** |
| 72 | 3 | Wola | **890 000** |
| 30 | 1 | Bemowo | **410 000** |

- **Wiersz** = jedna **obserwacja** (*instance / sample / row*) – tutaj jedno mieszkanie.
- Kolumny *metraż, liczba pokoi, dzielnica* = **cechy** (*features*). W kodzie oznaczamy je zwykle literą `X`.
- Kolumna *cena* = **zmienna celu** albo **etykieta** (*target / label*) – to, co chcemy przewidzieć. Oznaczamy ją `y`.
- **Model** – „przepis” wyuczony z danych, który na podstawie `X` przewiduje `y`.
- **Trening** (*training*) – proces uczenia modelu na danych.
- **Predykcja** (*prediction*) – użycie gotowego modelu na **nowych** danych.

❓ *Pytanie kontrolne:* czy przewidywanie ceny mieszkania to regresja, czy klasyfikacja? (Odpowiedź: **regresja**, bo cena to liczba.)

## Czym jest PyCaret?
**PyCaret** to biblioteka Pythona typu *low-code* („mało kodu”). Rzeczy, które normalnie wymagają dziesiątek linijek kodu
w bibliotece scikit-learn, w PyCaret robimy **jednym poleceniem**.

🚗 Analogia: PyCaret to **automatyczna skrzynia biegów**. Nadal prowadzisz samochód i decydujesz, dokąd jedziesz,
ale nie musisz ręcznie wrzucać każdego biegu.

Przez cały semestr będziemy powtarzać ten sam „przepis” (zapamiętaj go!):

```
setup()  →  compare_models()  →  create_model()  →  tune_model()  →  predict_model()  →  finalize_model()  →  save_model()
przygotuj    porównaj modele     zbuduj model       ulepsz model     przewiduj           trenuj na całości      zapisz
```

# 📝 Podsumowanie
- **Uczenie maszynowe** = komputer uczy się reguł z **przykładów** (danych), a nie z ręcznie napisanego kodu.
- **Uczenie nadzorowane**: znamy poprawne odpowiedzi → **regresja** (liczba) lub **klasyfikacja** (kategoria).
- **Uczenie nienadzorowane**: brak odpowiedzi → **grupowanie** lub **wykrywanie anomalii**.
- Tabela danych: wiersze = **obserwacje**, kolumny = **cechy** (`X`) + **zmienna celu** (`y`).
- `get_data()` z PyCaret wczytuje gotowe zbiory danych jednym poleceniem.

## 📖 Słowniczek PL / EN
| Polski | English | Znaczenie |
|---|---|---|
| uczenie maszynowe | Machine Learning (ML) | uczenie komputera na przykładach |
| uczenie nadzorowane | supervised learning | dane mają poprawne odpowiedzi |
| uczenie nienadzorowane | unsupervised learning | dane bez odpowiedzi |
| regresja | regression | przewidywanie liczby |
| klasyfikacja | classification | przewidywanie kategorii |
| grupowanie | clustering | dzielenie danych na podobne grupy |
| obserwacja | instance / sample | jeden wiersz tabeli |
| cecha | feature | kolumna opisująca obserwację (`X`) |
| zmienna celu / etykieta | target / label | to, co przewidujemy (`y`) |
| model | model | „przepis” wyuczony z danych |
| zbiór danych | dataset | tabela z danymi |
