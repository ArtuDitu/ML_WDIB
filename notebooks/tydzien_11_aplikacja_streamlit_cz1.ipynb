# 📚 Część 1: Wprowadzenie teoretyczne

## Po co aplikacja? 🤔
Model zapisany w pliku `.pkl` to jak **świetny przepis schowany w szufladzie** – nikt poza nami z niego nie skorzysta.
Żeby agent ubezpieczeniowy, lekarz czy klient mógł użyć modelu, trzeba go „podać na talerzu”: jako **aplikację internetową**
albo **API**. Ten proces nazywamy **wdrożeniem** (*deployment*).

W Pythonie są do tego różne narzędzia: **Django** i **Flask** (pełne frameworki webowe), **FastAPI** (do budowy API)…
My użyjemy **Streamlit** – najprostszego narzędzia do aplikacji z danymi i ML.

## Jak działa Streamlit? 🎈
- Piszesz **zwykły skrypt Pythona** (plik `.py`) – bez HTML, CSS i JavaScriptu.
- Polecenie `streamlit run plik.py` zamienia go w **stronę internetową**.
- Każda funkcja `st.coś()` dodaje element na stronie, **w kolejności od góry do dołu**.
- 🔁 **Najważniejsza zasada:** po każdej zmianie w widżecie (np. przesunięciu suwaka) Streamlit **uruchamia cały skrypt od nowa**,
  od pierwszej do ostatniej linii. To trochę jak odświeżenie strony.

## Najważniejsze widżety (*widgets*) 🧰
| Funkcja | Wygląd | Do czego? |
|---|---|---|
| `st.number_input` | pole z liczbą i przyciskami − / + | wiek, BMI |
| `st.radio` | przyciski opcji ◉ ○ | wybór jednej z kilku opcji (płeć) |
| `st.slider` | suwak | liczba z zakresu (liczba dzieci) |
| `st.selectbox` | lista rozwijana ▾ | wybór z dłuższej listy (region) |
| `st.checkbox` | pole wyboru ☑ | tak / nie (palenie) |
| `st.button` | przycisk | uruchomienie akcji |

## Formularz (*form*) 📋
Bez formularza każda zmiana od razu uruchamia skrypt. **Formularz** zbiera wszystkie odpowiedzi i wysyła je razem
dopiero po kliknięciu przycisku – jak papierowy formularz, który oddajesz w okienku dopiero po wypełnieniu całości.

## Jak będziemy pracować? 🛠️
Aplikacja Streamlit to **osobny plik `.py`**, a nie notebook. Nie musimy jednak przełączać się do edytora:
polecenie magiczne Jupytera **`%%writefile nazwa_pliku`** na początku komórki **zapisuje treść komórki do pliku**
(zamiast ją wykonywać). Wariant **`%%writefile -a`** **dopisuje** (*append*) treść na końcu pliku – dzięki temu
zbudujemy aplikację z małych kawałków.

### Krok 4: Budujemy kalkulator kosztów ubezpieczenia – kawałek po kawałku 🧱
Zbudujemy formularz z polami odpowiadającymi kolumnom zbioru `insurance`: `age`, `sex`, `bmi`, `children`, `region`, `smoker`.

👉 **Zostaw aplikację uruchomioną** i po każdym kroku kliknij **Rerun** na stronie – zobaczysz, jak aplikacja rośnie!
Najpierw zatrzymaj `hello.py` (<kbd>Ctrl</kbd> + <kbd>C</kbd>) i uruchom nowy plik:
```bash
streamlit run aplikacja/app.py
```
(Uruchom terminalowe polecenie **po** wykonaniu komórki 4a – wtedy plik już istnieje.)

#### 4a. Nagłówek i opis (`%%writefile` – tworzy plik od nowa)

````python
%%writefile ../aplikacja/app.py
# ============================================================
#  Kalkulator kosztów ubezpieczenia – aplikacja Streamlit
# ============================================================
import pandas as pd
import streamlit as st

# Ustawienia strony (tytuł karty przeglądarki) – musi być PIERWSZYM poleceniem st
st.set_page_config(page_title="Kalkulator kosztów ubezpieczenia")

st.title("💰 Kalkulator kosztów ubezpieczenia")
st.markdown("Wpisz swoje dane, aby poznać przewidywany roczny koszt ubezpieczenia zdrowotnego. "
            "Aplikacja powstała na kursie *Praktyczny Machine Learning z PyCaret*.")
````

#### 4b. Formularz: wiek i płeć (`%%writefile -a` – dopisuje na końcu pliku)
⚠️ Dopóki nie dodamy przycisku (krok 4e), Streamlit pokaże komunikat **„Missing Submit Button”** – to normalne!

````python
%%writefile -a ../aplikacja/app.py

# --- Formularz: widżety zbierają dane, ale wysyłamy je dopiero przyciskiem ---
form = st.form("charges")
age = form.number_input("Wiek", min_value=18, max_value=64, value=30)      # wiek w latach
sex = form.radio("Płeć", ["Mężczyzna", "Kobieta"])                         # przyciski opcji
````

💡 **Dlaczego wiek 18–64?** Model z tygodnia 4 uczył się tylko na osobach w tym wieku. Modele ML **słabo radzą sobie poza zakresem
danych treningowych** (nie umieją „ekstrapolować”), więc lepiej od razu ograniczyć możliwe wartości.

#### 4c. BMI i liczba dzieci

````python
%%writefile -a ../aplikacja/app.py
bmi = form.number_input("BMI (wskaźnik masy ciała)", min_value=15.0, max_value=55.0, value=25.0, step=0.1)
children = form.slider("Liczba dzieci", min_value=0, max_value=5, value=0)   # suwak 0–5
````

#### 4d. Region i palenie

````python
%%writefile -a ../aplikacja/app.py
region = form.selectbox("Region USA", ["southwest", "southeast", "northwest", "northeast"])   # lista rozwijana
smoker = form.checkbox("Palę papierosy")                                                       # True / False
````

#### 4e. Przycisk wysyłający formularz
`form_submit_button` zwraca `True` **tylko** w tym przebiegu skryptu, który nastąpił po kliknięciu przycisku.

````python
%%writefile -a ../aplikacja/app.py
submitted = form.form_submit_button("Pokaż moje dane")
````

#### 4f. Zamiana odpowiedzi na tabelę dla modelu 🔄
Model rozumie dane **dokładnie w takim formacie, jak w zbiorze treningowym**: `sex` = `'male'`/`'female'`,
`smoker` = `'yes'`/`'no'`. Musimy więc „przetłumaczyć” odpowiedzi z interfejsu na język danych.

````python
%%writefile -a ../aplikacja/app.py

# --- Tłumaczymy odpowiedzi z formularza na format danych treningowych (jeden wiersz tabeli) ---
input_df = pd.DataFrame([{
    'age': age,
    'sex': 'male' if sex == 'Mężczyzna' else 'female',
    'bmi': bmi,
    'children': children,
    'smoker': 'yes' if smoker else 'no',
    'region': region,
}])

# --- Po kliknięciu przycisku pokazujemy dane (za tydzień: przewidywany koszt!) ---
if submitted:
    st.success("Dziękujemy! Oto Twoje dane w formacie, który rozumie model:")
    st.dataframe(input_df)
````

### Krok 5: Sprawdzamy gotowy plik 🔍
Podejrzyjmy całą zawartość pliku `app.py`, żeby upewnić się, że wszystkie kawałki są na miejscu (i nic się nie zdublowało).

````python
# open(...).read() wczytuje cały plik jako tekst, a print go wyświetla
print(open('../aplikacja/app.py', encoding='utf-8').read())
````

⚠️ **Coś się zdublowało?** To się zdarza, gdy uruchomisz komórkę z `%%writefile -a` dwa razy. Rozwiązanie: uruchom ponownie
komórkę **4a** (z `%%writefile` bez `-a` – tworzy plik od nowa), a potem kolejne komórki 4b–4f **po jednym razie**.
